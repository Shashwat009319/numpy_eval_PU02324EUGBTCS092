```python
import numpy as np

np.random.seed(42)

A = np.random.randint(1, 11, size=(4, 3))
B = np.random.randint(1, 11, size=(3, 5))

print("Matrix A:")
print(A)
print("Shape:", A.shape)

print("\nMatrix B:")
print(B)
print("Shape:", B.shape)

assert A.shape[1] == B.shape[0]

C = np.dot(A, B)

print("\nMatrix C:")
print(C)
print("Shape:", C.shape)

print("\nTotal Sum:")
print(np.sum(C))

print("\nRow-wise Sum:")
print(np.sum(C, axis=1))

print("\nColumn-wise Sum:")
print(np.sum(C, axis=0))
```
